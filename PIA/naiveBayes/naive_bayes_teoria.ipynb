# Teoria praktikoa: Naive Bayes

Naive Bayes, Bayesen teoreman oinarritutako modelo probabilistikoa da. Hurrengo funtzioan horrela defini daitekeena:

$P(A \mid B) = \frac{P(A \cap B)}{P(B)} = \frac{P(B \mid A) \cdot P(A)}{P(B)}$

Non, $P(A \mid B)$  $A$ okurritzeko probabilitatea $B$ okurritu dela jakinda esan nahi du.
